## Section 1 — Visualization Answers Questions

### Core idea

Data visualization is not just about making charts.

**The point of a visualization is to help answer a question.**

A good chart should make a pattern easier to see. A weak chart may technically show the data, but it can make the pattern harder to understand.

---

### Objectives

By the end of this section, students should understand:
* Different chart types answer different kinds of questions.
* The first step is not “which code do I write?”
* The first step is “what question am I trying to answer?”

## Chart Selection Framework

| Data Situation                  | Question                                       | Chart Type            |
| ------------------------------- | ---------------------------------------------- | --------------------- |
| One numeric column              | What does the distribution look like?          | Histogram             |
| One categorical column          | How many observations are in each group?       | Bar chart / countplot |
| Two numeric columns             | Are these variables related?                   | Scatterplot           |
| One numeric + one categorical   | How does the numeric variable differ by group? | Boxplot               |
| Ordered x-axis + numeric y-axis | How does the value change across order/time?   | Line plot             |
| Several numeric columns         | Which variables move together?                 | Heatmap               |
| Multiple groups                 | How does the pattern differ by group?          | Hue / subplots        |

### How to read a boxplot:

Boxplots are considered more technical visualizations, but they're essentially showing the distribution with respect to a category.

| Boxplot Part         | Meaning                |
| -------------------- | ---------------------- |
| Line inside the box  | Median                 |
| Bottom of the box    | 25th percentile        |
| Top of the box       | 75th percentile        |
| Height of the box    | Middle 50% of values   |
| Whiskers             | Typical low/high range |
| Dots beyond whiskers | Possible outliers      |


A common mistake is using bar charts for all group comparisons

This isn't incorrect but it hides the spread of the data.

A boxplot shows a lot more information.

---

**An Aside**

Beyond the boxplot is the strip plot, which shows individual observations.

````python
plt.figure(figsize=(8, 5))

sns.boxplot(data=mpg, x="origin", y="mpg")
sns.stripplot(data=mpg, x="origin", y="mpg", alpha=0.4)

plt.title("Miles Per Gallon by Origin")
plt.xlabel("Origin")
plt.ylabel("Miles Per Gallon")

plt.show()
````

### Heatmaps

A heatmap can show relationships among several numeric variables.

For this section, use a correlation heatmap.

Guiding question:

> Which numeric variables are most strongly related to each other?

````python
# First grab the numeric columns
numeric_cols = [
    "mpg",
    "cylinders",
    "displacement",
    "horsepower",
    "weight",
    "acceleration",
    "model_year"
]

mpg_corr = mpg[numeric_cols].corr()

mpg_corr
````

````python
# and plot our heatmap
plt.figure(figsize=(10, 6))

sns.heatmap(mpg_corr, annot=True)

plt.title("Correlation Heatmap of Numeric MPG Variables")

plt.show()
````

````python
# We can also change the seaborn color map with cmap
plt.figure(figsize=(10, 6))

sns.heatmap(mpg_corr, annot=True, cmap="coolwarm", center=0)

plt.title("Correlation Heatmap of Numeric MPG Variables")

plt.show()
````

````python
plt.figure(figsize=(10, 6))

sns.heatmap(mpg_corr, annot=True, cmap="bone", center=0)

plt.title("Correlation Heatmap of Numeric MPG Variables")

plt.show()
````
